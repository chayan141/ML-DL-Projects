# Self- Supervised Learning:

Self supervised learning : There is no predefined labels, The model uses the structure or context within the data itself to generate pseudo-labels for training.

In NLP, the model might predict masked words in a sentence, using the surrounding words as context.

Based on context, model automatically generates pseudo-labels.

Prompt Techniques: Few Shot, Zero Shot, Chain of Throught, Prompt chaining (output of one prompt has fed into next prompt), Template prompting (predefined prompt structure), Instruction based prompting, Role play prompting, Self consistency prompting: The model generates multiple response to the same prompt and selects the most consistent prompt.

# Cross Modal Learning

Its a machine learning approach where a model learns to understand and integrate information from mutiple modalities, such as text, images, audio or video.

Example Image Captioning, Cross Modal Retrival.

Image Captioning: CLIP (Contrasive Image text pretraining) a model is trained on img text pairs to associate images with their captions. It learns to map both modalities into a shared embedding space.

Data Collection : COCO Dataset.

Preprocessing: Tokenized and Embedded.

Images are Processed with VIT or CNNs to extract visual features.

Audios are converted into spectogram using wav2vec.

Videos are extracted by 3D CNNs.

Cross Modal Attention : Uses attention mechenisms to allow modalities dynamically to relevent parts of each others. (Models like ViBERT or CLIP).

Map features from all modalities into a common embedding space where related concepts are closed to gether.

Contrasive Learning: Encourages paired data e.g. image and its caption to have similar representation while pushing unrelated pairs. CLIP uses contrasive loss to align image and its caption.

Self Supervised : Masks parts one modality and predict them using other.

Dalle or stable diffuion: Uses text imagepairs to generate image from text prompt

Models like clip can generalize to new tasks without fine-tuning due to robust pretraining.

Contrasive Learning: Model learns to distinguish between similar and dissimilar pairs of data point by optimizing a loss function that brings similar pairs closer together and dissimilar pairs aparts.

The goal is to learn an embedding space where related data points are mapped together while unrelated ones are far apart.

In contrasive learning, models learn to project multimodal data into shared embedding space where similar pairs are closer together.

Zero Shot : Model can classify the image into its respective caption which it has not been trained on.

Diffusion model used to denoise predicted images.
