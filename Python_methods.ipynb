# String Indexing and slicing

Indexing

Positive indices: 0 (first character) to len(string) - 1 (last character).

Negative indices: -1 (last character) to -len(string) (first character).

Slicing

Syntax: string[start:end:step]

start: The index where the slice begins (inclusive). Defaults to 0 if omitted.

end: The index where the slice ends (exclusive). Defaults to len(string) if omitted.

step: The increment between indices (default is 1). A negative step reverses the direction.

Omitting start or end uses the string’s start or end, respectively.

A negative step reverses the string or substring.

# Shallow Copy vs Deep Copy

Shallow Copy: The top level elements are copied but nested objects are still shared between the original and the copy

Use the copy() method, list() constructor, or copy.copy() from the copy module.

```python
import copy
original = [1, 2, [3, 4]]

shallow = copy.copy(original)
shallow[0] = 10
print(original)
print(shallow)
```

```python
shallow[2][0] = 7
print(original)
print(shallow)
```

Deepcopy: creates a completely independent copy of the object and all nested objects recursively. No reference are shared between the original and the copy.

Use copy.deepcopy() from the copy module.

```python
import copy

original = [1, 2, [3, 4]]
deep = copy.deepcopy(original)  # Deep copy

# Modify nested list
deep[2][0] = 30
print(original)  # Output: [1, 2, [3, 4]]  (original unchanged)
print(deep)      # Output: [10, 2, [30, 4]]
```

# Tuple Methods

Tuples are immutable, ordered collection of elements that can store heterogeneous data (numbers, strings or other objects). They are similar to lists but can not be modified after creation (immutable)

List : Ordered mutable collection
* Mutable : Can modify elements (add, remove).
* Ordered (Index-based access)
* Duplicates : It allows duplicate
* Elements can be any type.

Set : Unordered (indexing can not be done), mutable collection of unique elements

* Mutable : Can modify elements (add, remove).
* Unordered (No indexing or slicing)
* No duplicates: Only unique elements
* Immutable Types only (must have same datatype)

Tuples: Ordered (indexing can be done) immutable (no modification of elements)

* Immutable : Elements cant be modified.
* Ordered (Index Based access)
* It allows duplicate
* Can have any datatype.

Dictionary: Unordered (No index based access, only key based access), mutable (can be modified) collection of key value pair.

* Mutable : Can be modified. (Adds or remove)
* Unordered (No indexing, only key based access)
* Keys must be unique but values can be duplicated.
* Keys are immutable (type (should be same datatype) but values can be any type.

```python
# List
my_list = [1, 2, 2, "apple"]
my_list.append(3)  # Mutable
print(my_list)  # Output: [1, 2, 2, "apple", 3]
```

# Json Module

It allows you to serialize python objects into json format and deserialize json dat into python objects.

JSONs widely used for data exchange, especially in APIs and configuration files.

It converts python objects like dictionary, lists, tuples into json string and vice versa.

Serialization: Python objects to JSON. (json.dump)

Deseralization: Deserialize JSON data from a file like object into a python object.

```python
#Serialization
import json

# Python dictionary
data = {
    "name": "Alice",
    "age": 30,
    "scores": [85, 90, 95],
    "active": True,
    "address": {"city": "New York", "zip": None}
}

json_string = json.dumps(data, indent=2, sort_keys=True)
print(json_string)
```

Output (formatted with indent=2):

```python
import json

with open("data.json","w") as file:
  json.dump(data, file, indent=2)
```

```python
#Deserialization
data = json.loads(json_string)
print(data)
```

# Itertools Functions

Provides iterator building tools for efficient looping, combination and permutation tasks.

```python
from itertools import count
for i in count(10,2):
  if i > 16:
    break
  print(i)
```

```python
from itertools import cycle

colors = ['red', 'blue', 'green']
counter = 0
for color in cycle(colors):
    if counter >= 6:
        break
    print(color)  # Output: red, blue, green, red, blue, green
    counter += 1
```

```python
l = [1,2,3,4,5]
a = []

for i in range(len(l)):
  if i<len(l)-1:
    a.append(l[i]+l[i+1])
print(a)
```

```python
# prompt: calculate running sum of list elements

def running_sum(numbers):
  """Calculates the running sum of a list of numbers."""
  sums = []
  current_sum = 0
  for number in numbers:
    current_sum += number
    sums.append(current_sum)
  return sums

result = running_sum(a)
print(result)

# Using a list comprehension
running_sums_comp = [sum(a[:i+1]) for i in range(len(a))]
running_sums_comp
```

Iterators : An iterator is an object that implements the iterator protocol, which consists of the __iter__() and __next__() methods. It allows sequential access of elements.

__iter __(): Returns iterator object itself.

__next __(): Returns the next element to stop iterations.

```python
my_list = [1, 2, 3]
```

```python
iterator = iter(my_list)
next(iterator)
```

Generators : A special type of iterator created using a function with yield statement or generator expression. It generates the value lazily , one at a time.

It is extremely memory efficient, as it generates the value lazily, generates on demand without storing entire sequence.

```python
def my_gen():
  yield 1
  yield 2
  yield 3

gen = my_gen()

next(gen)
```

```python
gen = (x*x for x in range(3))
print(list(gen))
```

```python
class Counter:
  def __init__(self, start, end):
    self.current = start
    self.end = end

  def __iter__(self):
    return self

  def __next__(self):
    if self.current > self.end:
      raise StopIteration

    value = self.current
    self.current +=1
    return value

counter = Counter(1,3)
print(list(counter))
```

```python
def fibonacci(n):
  a = 0
  b = 1

  for i in range(n):
    yield a
    a, b = b, a + b

print(list(fibonacci(5)))  # Output: [0, 1, 1, 2, 3]
```
