A logger is a component of Python’s built-in logging module that lets you:

Track events that happen during execution

Record debug messages, warnings, errors, etc.

Write logs to console, files, or external systems

It's better than using print() because:

It provides log levels (DEBUG, INFO, WARNING, ERROR, CRITICAL)

Logs can be saved to a file

You can format logs with timestamps, line numbers, etc.

You can control the log output via configuration

# Basic Logger

```python
import logging

# Configure logger
logging.basicConfig(
    level=logging.DEBUG,
    format='%(asctime)s - %(levelname)s - %(message)s'
)

# Logging messages of different severity
logging.debug("This is a debug message")
logging.info("This is an info message")
logging.warning("Warning! Something is off.")
logging.error("An error occurred")
logging.critical("Critical issue!")
```

```python
logging.basicConfig(
    filename='app.log',
    level=logging.INFO,
    format='%(asctime)s - %(levelname)s - %(message)s'
)
logging.info("App started")
```

## Custom Logger

```python
import logging

logger = logging.getLogger(__name__)
logger.setLevel(logging.INFO)

handler = logging.FileHandler('module.log')
formatter = logging.Formatter('%(name)s - %(levelname)s - %(message)s')
handler.setFormatter(formatter)
logger.addHandler(handler)

def divide(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        logger.error("Division by zero attempted!")

divide(5, 0)
```

# iamgetter function in Python

itemgetter() is a function from the operator module that creates a callable (function) to fetch an item or multiple items from a sequence (like a list, tuple, or string) by index or from a dictionary by key.

```python
from operator import itemgetter

data = [('Alice', 25), ('Bob', 20), ('Charlie', 30)]

# Sort by age (2nd item, index 1)
sorted_data = sorted(data, key=itemgetter(1))

print(sorted_data)
```

```python
from operator import itemgetter

record = ('John', 'Doe', 35, 'Engineer')

# Get first and last name
get_name = itemgetter(0, 1)
print(get_name(record))  # ➝ ('John', 'Doe')
```

```python
from operator import itemgetter

people = [
    {'name': 'Alice', 'age': 25},
    {'name': 'Charlie', 'age': 30},
    {'name': 'Bob', 'age': 20}
]

# Sort by 'name'
sorted_people = sorted(people, key=itemgetter('name'))

print(sorted_people)
```
