# Basic Logger

```python
import logging

# Configure logger
logging.basicConfig(
    level=logging.DEBUG,
    format='%(asctime)s - %(levelname)s - %(message)s'
)

# Logging messages of different severity
logging.debug("This is a debug message")
logging.info("This is an info message")
logging.warning("Warning! Something is off.")
logging.error("An error occurred")
logging.critical("Critical issue!")
```

# Decorators

A decorator is a callable (usually a function) that takes another function as an argument and returns a new function that typically wraps or enhances the original function's behavior.

Staticmethod: Declares a method in a class as a static method, which does not require an instance (self) or class (cls) reference. It behaves like a regular function but belongs to the class's namespace.

It can also be called on an instance.

```python
class MathUtils:
    @staticmethod
    def add(a, b):
        return a + b

# Call without instantiating the class
result = MathUtils.add(3, 5)
print(result)  # Output: 8

# Can also be called on an instance, but no 'self' is passed
obj = MathUtils()
print(obj.add(2, 4))  # Output: 6
```

Classmethod: Purpose: Declares a method that receives the class itself (cls) as its first argument instead of an instance (self). It can access and modify class-level data.

```python
class Employee:
    company = "TechCorp"  # Class attribute

    def __init__(self, name):
        self.name = name

    @classmethod
    def from_string(cls, employee_str):
        name, company = employee_str.split("-")
        emp = cls(name)  # Create instance using cls
        emp.company = company  # Modify class attribute or instance
        return emp

    # Create instance using classmethod
emp = Employee.from_string("Alice-Google")
print(emp.name, emp.company)  # Output: Alice Google
print(Employee.company)  # Output: TechCorp
```

Propertymethod: Allows a method to be accessed like an attribute, enabling getter, setter, and deleter functionality for controlled access to instance attributes.

Used for Encapsulating attribute access to add logic (e.g., validation) without changing the interface.

```python
class Circle:
    def __init__(self, radius):
        self._radius = radius

    @property
    def radius(self):
      return self._radius

    @radius.setter
    def radius(self,value):
      if value <=0:
        raise ValueError("Must be positive")

      self._radius = value

    @property
    def area(self):
      return 3.14*self._radius**2

    @radius.deleter
    def radius(self):
      print("Deleting radius")
      del self._radius
```

```python
# Usage
circle = Circle(5)
print(circle.radius)  # Output: 5 (calls getter)
```

```python
print(circle.area)  # Output: 78.53975 (computed property)
circle.radius = 10  # Calls setter
```

```python
print(circle.radius)  # Output: 10
# circle.radius = -1  # Raises ValueError
del circle.radius  # Calls deleter, prints "Deleting radius"
```

```python
from functools import wraps

#preserves additional metadata

def my_decorator(func):
    @wraps(func)  # Preserves func's metadata
    def wrapper(*args, **kwargs):
        print(f"Calling {func.__name__}")
        return func(*args, **kwargs)
    return wrapper

@my_decorator
def greet(name):
    """Greets the user."""
    return f"Hello, {name}!"

print(greet("Alice"))  # Output: Calling greet\nHello, Alice!
print(greet.__name__)  # Output: greet (preserved by @wraps)
print(greet.__doc__)   # Output: Greets the user.
```

```python
def bad_decorator(func):
    def wrapper(*args, **kwargs):
        print(f"Calling {func.__name__}")
        return func(*args, **kwargs)
    return wrapper

@bad_decorator
def greet(name):
    """Greets the user."""
    return f"Hello, {name}!"

print(greet.__name__)  # Output: wrapper (metadata lost)
print(greet.__doc__)   # Output: None
```

lru_cache: Used for speeding up expensive or repetative function calls.

Used for functions with deterministic outputs and costly computations.

```python
from functools import lru_cache

@lru_cache(maxsize=128)
def fibonacci(n):
  if n <=1:
    return n
  return fibonacci(n-1)+fibonacci(n-2)

print(fibonacci(50))  # Output: 12586269025 (computed quickly due to caching)
print(fibonacci.cache_info())  # Output: CacheInfo(hits=48, misses=51, maxsize=128, currsize=51)
```

@lru_cache stores results for each unique input n, so subsequent calls with the same n retrieve the cached result.

The maxsize parameter limits the cache size; None means unlimited.

# Python dataclass

Purpose: Simplifies class creation by automatically generating special methods like __init__, __repr__, __eq__, etc., for classes that primarily store data.

```python
from dataclasses import dataclass

@dataclass
class Point:
    x: int
    y: int
    z: int = 0  # Default value

# Automatically gets __init__, __repr__, __eq__, etc.
p1 = Point(1, 2)
p2 = Point(1, 2, 0)
print(p1)  # Output: Point(x=1, y=2, z=0)
print(p1 == p2)  # Output: True (auto-generated __eq__)
```

```python
class Point:
    def __init__(self, x, y, z=0):
        self.x = x
        self.y = y
        self.z = z

    def __repr__(self):
        return f"Point(x={self.x}, y={self.y}, z={self.z})"

    def __eq__(self, other):
        if not isinstance(other, Point):
            return False
        return self.x == other.x and self.y == other.y and self.z == other.z
```

# AbstractMethod

An method which is abstract in abstract base class, requireing subclasses to implement it. Part of ABC module.

It ensures subclasses implement specific methods.

```python
from abc import ABC, abstractmethod

class Animal(ABC):
    @abstractmethod
    def make_sound(self):
        pass

class Dog(Animal):
    def make_sound(self):
        return "Woof!"

class Cat(Animal):
    def make_sound(self):
        return "Meow!"

# Cannot instantiate Animal directly
# animal = Animal()  # Raises TypeError: Can't instantiate abstract class

dog = Dog()
print(dog.make_sound())  # Output: Woof!
cat = Cat()
print(cat.make_sound())  # Output: Meow!
```

# Generics

It is enhanced with typing  module. Generic allows developers to define functions, classes or data structures that can operate on any type while maintaining type safety when used with static type checkers like mypy, pyright or IDEs.

Generics are particularly useful for defining type-agnostic code, where the same logic applies to multiple types (e.g., lists of integers, strings, or custom objects) without sacrificing type checking.

TypeVar : Defines a type variable that can represent an type or a restricted set of types.

Generic : A base class used to create generic classes that work with specific types.

Parameterized Types: Types like List[T], Dict[K, V], or Tuple[T, ...] that specify the types of their elements.

* Generic ensures that code is type checked at development time (Using static type checkers), reducing runtime error.

* Write function and classes once and use them with different types without duplicating code.

# Generic Functions with TypeVar

Write a function that returns the first element of a sequence, working with any type.

```python
from typing import TypeVar, Sequence

T = TypeVar('T')

def get_first(seq: Sequence[T]) -> T:
  return seq[0]

# Usage
numbers = [1, 2, 3]
strings = ["apple", "banana", "cherry"]
print(get_first(numbers))  # Output: 1 (inferred as int)
print(get_first(strings))  # Output: apple (inferred as str)
```

T is a type variable that can represent any type (e.g., int, str).

Sequence[T] indicates the input is a sequence of type T (e.g., List[int], Tuple[str]).

# Generic Classes with Generic

Create a generic Stack class that can store elements of any type.

```python
from typing import TypeVar, Generic, List

T = TypeVar('T')

class Stack(Generic[T]):
  def __init__(self):
    self.items: List[T] = []

  def push(self, item: T) -> None:
    self.items.append(item)

  def pop(self) -> T:
    return self.items.pop()

# Usage
stack = Stack[int]()
stack.push(10)
stack.push(20)
print(stack.pop())
```

```python
str_stack = Stack[str]()  # Stack of strings
str_stack.push("hello")
print(str_stack.pop())  # Output: hello
```

Stack(Generic[T]) declares Stack as a generic class parameterized by T.

The type of items is List[T], and methods push and pop use T consistently.

# Constrained Type Variables

Write a function that adds two values, restricting the type to numbers (int or float).

```python
from typing import TypeVar, Union

Number = TypeVar('Number', int, float)  # Constrain to int or float

def add(a: Number, b: Number) -> Number:
    return a + b

# Usage
print(add(1, 2))      # Output: 3 (int)
print(add(1.5, 2.5))  # Output: 4.0 (float)
# print(add("a", "b"))  # Type checker error: str not compatible with Number
```

# Generic Classes with Multiple Type Variable

Create a generic Pair class to store two values of potentially different types.

```python
from typing import TypeVar, Generic

K = TypeVar('K')
V = TypeVar('V')

class Pair(Generic[K, V]):
    def __init__(self, key: K, value: V):
        self.key = key
        self.value = value

    def get_key(self) -> K:
        return self.key

    def get_value(self) -> V:
        return self.value

# Usage
pair = Pair[str, int]("age", 30)
print(pair.get_key())   # Output: age
print(pair.get_value()) # Output: 30

# Type checker will catch errors
# pair = Pair[str, int](42, "invalid")  # Error: Incompatible types
```

Define a generic function that works with any iterable type and returns its length.

```python
from typing import TypeVar, Protocol, Iterable

T = TypeVar('T')

class Sized(Protocol):
    def __len__(self) -> int:
        ...

def count_elements(items: Iterable[T]) -> int:
    return len(items)

# Usage
print(count_elements([1, 2, 3]))        # Output: 3
print(count_elements(("a", "b")))       # Output: 2
print(count_elements({1: "one", 2: "two"}))  # Output: 2
```

Sized is a protocol ensuring the input supports __len__.

Iterable[T] allows the function to accept any iterable (list, tuple, dict, etc.) while preserving type information for its elements.
